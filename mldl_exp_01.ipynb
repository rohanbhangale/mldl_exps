```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split

# Load the diabetes dataset
diabetes = load_diabetes()
X_linear = diabetes.data
y_linear = diabetes.target

print(f"Shape of X_linear: {X_linear.shape}")
print(f"Shape of y_linear: {y_linear.shape}")

# Split the data into training and testing sets
X_train_linear, X_test_linear, y_train_linear, y_test_linear = train_test_split(
    X_linear, y_linear, test_size=0.2, random_state=42
)

print(f"Shape of X_train_linear: {X_train_linear.shape}")
print(f"Shape of X_test_linear: {X_test_linear.shape}")
print(f"Shape of y_train_linear: {y_train_linear.shape}")
print(f"Shape of y_test_linear: {y_test_linear.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Instantiate a Linear Regression model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train_linear, y_train_linear)

# Make predictions on the test set
y_pred_linear = linear_model.predict(X_test_linear)

# Calculate evaluation metrics
mse_linear = mean_squared_error(y_test_linear, y_pred_linear)
r2_linear = r2_score(y_test_linear, y_pred_linear)

# Print the metrics
print(f"Linear Regression - Mean Squared Error (MSE): {mse_linear:.2f}")
print(f"Linear Regression - R-squared (R2): {r2_linear:.2f}")
```

## Visualize Linear Regression Results

### Subtask:
Generate a scatter plot comparing actual vs. predicted values for linear regression, and a residuals plot to analyze the model's performance. Ensure legends are provided.


```python
import matplotlib.pyplot as plt

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Actual vs. Predicted Values
axes[0].scatter(y_test_linear, y_pred_linear, alpha=0.6, label='Actual vs. Predicted')
axes[0].plot([y_test_linear.min(), y_test_linear.max()], [y_test_linear.min(), y_test_linear.max()], 'r--', lw=2, label='Perfect Prediction')
axes[0].set_xlabel('Actual Values')
axes[0].set_ylabel('Predicted Values')
axes[0].set_title('Linear Regression: Actual vs. Predicted Values')
axes[0].legend()

# Plot 2: Residuals Plot
residuals_linear = y_test_linear - y_pred_linear
axes[1].scatter(y_pred_linear, residuals_linear, alpha=0.6)
axes[1].axhline(y=0, color='r', linestyle='--', lw=2)
axes[1].set_xlabel('Predicted Values')
axes[1].set_ylabel('Residuals')
axes[1].set_title('Linear Regression: Residuals Plot')

# Adjust layout and display plots
plt.tight_layout()
plt.show()
```

## Load and Prepare Dataset for Logistic Regression

### Subtask:
Load a suitable dataset for binary classification from `sklearn.datasets` (e.g., `load_breast_cancer`), explore its features, and prepare it for modeling by splitting into training and testing sets.


```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

# Load the breast cancer dataset
breast_cancer = load_breast_cancer()
X_logistic = breast_cancer.data
y_logistic = breast_cancer.target

print(f"Shape of X_logistic: {X_logistic.shape}")
print(f"Shape of y_logistic: {y_logistic.shape}")

# Split the data into training and testing sets
X_train_logistic, X_test_logistic, y_train_logistic, y_test_logistic = train_test_split(
    X_logistic, y_logistic, test_size=0.2, random_state=42
)

print(f"Shape of X_train_logistic: {X_train_logistic.shape}")
print(f"Shape of X_test_logistic: {X_test_logistic.shape}")
print(f"Shape of y_train_logistic: {y_train_logistic.shape}")
print(f"Shape of y_test_logistic: {y_test_logistic.shape}")
```

## Implement and Evaluate Logistic Regression

### Subtask:
Train a Logistic Regression model on the prepared dataset, make predictions on the test set, and evaluate its performance using metrics like accuracy, precision, recall, F1-score, and a confusion matrix.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler

# Instantiate a StandardScaler
scaler = StandardScaler()

# Scale the training and test data for logistic regression
X_train_logistic_scaled = scaler.fit_transform(X_train_logistic)
X_test_logistic_scaled = scaler.transform(X_test_logistic)

# Instantiate a Logistic Regression model
logistic_model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model using scaled data
logistic_model.fit(X_train_logistic_scaled, y_train_logistic)

# Make predictions on the scaled test set
y_pred_logistic = logistic_model.predict(X_test_logistic_scaled)

# Calculate evaluation metrics
accuracy_logistic = accuracy_score(y_test_logistic, y_pred_logistic)
precision_logistic = precision_score(y_test_logistic, y_pred_logistic)
recall_logistic = recall_score(y_test_logistic, y_pred_logistic)
f1_logistic = f1_score(y_test_logistic, y_pred_logistic)
conf_matrix_logistic = confusion_matrix(y_test_logistic, y_pred_logistic)

# Print the metrics
print(f"Logistic Regression - Accuracy: {accuracy_logistic:.2f}")
print(f"Logistic Regression - Precision: {precision_logistic:.2f}")
print(f"Logistic Regression - Recall: {recall_logistic:.2f}")
print(f"Logistic Regression - F1-Score: {f1_logistic:.2f}")
print("Logistic Regression - Confusion Matrix:")
print(conf_matrix_logistic)
```

## Visualize Logistic Regression Results and Heatmap

### Subtask:
Generate a heatmap for the confusion matrix of the Logistic Regression model. Also, generate an ROC curve to visualize the classifier's performance. Ensure legends are provided.


```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import RocCurveDisplay

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Confusion Matrix Heatmap
sns.heatmap(conf_matrix_logistic, annot=True, fmt='d', cmap='Blues', ax=axes[0])
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('True Label')
axes[0].set_title('Logistic Regression: Confusion Matrix')

# Plot 2: ROC Curve
RocCurveDisplay.from_estimator(logistic_model, X_test_logistic_scaled, y_test_logistic, ax=axes[1])
axes[1].set_title('Logistic Regression: ROC Curve')
axes[1].legend(loc='lower right')

# Adjust layout and display plots
plt.tight_layout()
plt.show()
```

## Hyperparameter Tuning for Logistic Regression

### Subtask:
Perform hyperparameter tuning for the Logistic Regression model using `GridSearchCV` to find the best parameters. We will tune the `C` parameter (inverse of regularization strength) and `solver` for the Logistic Regression model. Then, we will use the best model to make predictions and evaluate its performance.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report

# Define the parameter grid for GridSearchCV
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs']
}

# Instantiate a Logistic Regression model
# max_iter is increased to ensure convergence for different solvers
logistic_model_tuning = LogisticRegression(max_iter=1000, random_state=42)

# Instantiate GridSearchCV
grid_search = GridSearchCV(
    estimator=logistic_model_tuning,
    param_grid=param_grid,
    cv=5, # 5-fold cross-validation
    scoring='accuracy', # Evaluate based on accuracy
    n_jobs=-1, # Use all available cores
    verbose=1
)

# Fit GridSearchCV to the scaled training data
grid_search.fit(X_train_logistic_scaled, y_train_logistic)

# Print the best parameters and best score
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best cross-validation accuracy: {grid_search.best_score_:.2f}")

# Get the best model
best_logistic_model = grid_search.best_estimator_

# Make predictions with the best model
y_pred_best_logistic = best_logistic_model.predict(X_test_logistic_scaled)

# Evaluate the best model
accuracy_best_logistic = accuracy_score(y_test_logistic, y_pred_best_logistic)
precision_best_logistic = precision_score(y_test_logistic, y_pred_best_logistic)
recall_best_logistic = recall_score(y_test_logistic, y_pred_best_logistic)
f1_best_logistic = f1_score(y_test_logistic, y_pred_best_logistic)
conf_matrix_best_logistic = confusion_matrix(y_test_logistic, y_pred_best_logistic)

print(f"\nBest Logistic Regression - Accuracy: {accuracy_best_logistic:.2f}")
print(f"Best Logistic Regression - Precision: {precision_best_logistic:.2f}")
print(f"Best Logistic Regression - Recall: {recall_best_logistic:.2f}")
print(f"Best Logistic Regression - F1-Score: {f1_best_logistic:.2f}")
print("Best Logistic Regression - Confusion Matrix:")
print(conf_matrix_best_logistic)
```

## Visualize Hyperparameter Tuning Results

### Subtask:
Generate a graph to visualize the results of the hyperparameter tuning for the Logistic Regression model. This will typically show how the model's performance (e.g., accuracy) changes with different hyperparameter values.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert GridSearchCV results to a DataFrame for easier plotting
results_df = pd.DataFrame(grid_search.cv_results_)

# Filter for relevant columns
results_df = results_df[['param_C', 'param_solver', 'mean_test_score', 'std_test_score']]

# Rename columns for clarity
results_df.rename(columns={
    'param_C': 'C_value',
    'param_solver': 'Solver',
    'mean_test_score': 'Mean_Accuracy',
    'std_test_score': 'Std_Accuracy'
}, inplace=True)

# Create the plot
plt.figure(figsize=(10, 6))
sns.lineplot(data=results_df, x='C_value', y='Mean_Accuracy', hue='Solver', marker='o')

plt.xscale('log') # Use a log scale for C values as they span orders of magnitude
plt.xlabel('C Value (Inverse of Regularization Strength)')
plt.ylabel('Mean Cross-Validation Accuracy')
plt.title('Logistic Regression Hyperparameter Tuning: C vs. Accuracy')
plt.grid(True, which="both", ls="--", c='0.7')
plt.legend(title='Solver')
plt.tight_layout()
plt.show()
```
